# **How Quantum Mechanics is Applied to the Quantum Penalty Shootout Game** ⚽🥅:

In this game, quantum mechanics plays a crucial role in determining the goalkeeper's actions, creating an unpredictable and exciting element. Here's a breakdown of how quantum principles are applied:

## Qubits and Their Role in the Game

In the Quantum Penalty Shootout game, qubits play a central role in simulating the probabilistic nature of the goalkeeper’s decision-making process during the penalty shootout
 Here's a detailed explanation of how qubits are used in the context of the game:

## Superposition:

## Superposition (Creating Uncertainty)

What is superposition?

In quantum mechanics, a qubit can exist in a superposition of both |0⟩ and |1⟩ states at the same time, unlike classical bits which are either 0 or 1. This allows quantum systems to represent more information simultaneously.

How is it used in the game?

The quantum goalkeeper (represented by qubits) is placed in superposition using Hadamard gates. This means the goalkeeper’s possible actions (dive directions) are not determined ahead of time—they are a mix of multiple possibilities.In the game, the goalkeeper can have multiple possible dive positions at once before the actual measurement is made. The superposition represents all these possibilities being present simultaneously.

For example, applying a Hadamard gate to a qubit transforms it from |0⟩ to a superposition state:

$$ |0\rangle \rightarrow \frac{1}{\sqrt{2}} \left( |0\rangle + |1\rangle \right) $$

This superposition creates uncertainty in the goalkeeper’s actions and mimics the randomness of real-world penalty shootouts.

## Entanglement:

**How it's used**:

Entanglement is created using the CNOT gates applied between pairs of qubits. This means the states of certain qubits are dependent on the states of others, such that the outcome of one qubit’s measurement directly influences the others.

For instance, if one qubit determines a "Top Left" dive, the entangled qubit will reflect a corresponding outcome that influences the goalkeeper’s response. The goalkeeper’s decision is thus entangled with the shot direction the qubit represents.

This entanglement adds a level of complexity and quantum unpredictability, as the goalkeeper’s dive may be influenced by the entangled quantum states, simulating a real-world decision-making process that may not always be deterministic.

## Measurement:

**How it's used**:

After the qubits are put into superposition and entangled, they are measured. The measurement process collapses the qubits' state to one of the possible outcomes, such as "Top Left," "Bottom Left," "Center," or "Top Right."

The measurement determines the final decision made by the goalkeeper (whether they dive to the left, center, or right), and the outcome influences whether the penalty shot results in a goal, a save, or a miss.

The randomness of the measurement is what makes the game unpredictable. Just like in real-life penalties, sometimes the goalkeeper "guesses" correctly, and other times they don't.

## Probabilistic Outcomes:

**How it's used**:

The probabilistic nature of quantum measurement ensures that the goalkeeper’s response is not deterministic. Instead of always diving in the same direction based on prior knowledge, the goalkeeper’s dive is probabilistically determined by the state of the qubits, resulting in a random and dynamic penalty shootout.

When the qubits are measured, the outcome is not set until the measurement is made, just like a quantum computer performs operations that aren’t fully determined until observed.

## How Qubits Drive the Gameplay:
Quantum Result Generation:

The game generates the goalkeeper’s dive decision using quantum mechanics, specifically via the superposition and entanglement of qubits. Each time a penalty is taken, the qubits are manipulated using quantum gates to simulate different potential dive outcomes. The final decision is made when the qubits are measured.

## Mapping to Shot Directions:

The quantum results (bitstrings like "00", "01", "10", "11") are mapped to specific shot directions (Top Left, Bottom Left, Center, Top Right). This mapping ensures that the outcome of the goalkeeper’s dive (determined by quantum mechanics) aligns with one of these directions in the game.

## Randomness and Uncertainty:

The qubits introduce randomness and uncertainty into the game. Just like a real-world penalty, where the goalkeeper cannot predict with certainty where the ball will be shot, the qubits' probabilistic nature ensures that each penalty has an unpredictable outcome. The goalkeeper's decision is based on the quantum results, leading to an exciting and dynamic gameplay experience.

##Simulating Real-Life Decision Making:

In a real penalty shootout, a goalkeeper's decision to dive in a particular direction is influenced by a variety of factors, including instinct, prior knowledge, or even randomness. In this game, the qubits mimic this process, with the quantum operations providing the randomness and unpredictability seen in actual penalty shootouts.

## 5)Quantum Result Generation Function

    get_quokka_result: This function sends a quantum circuit written in QASM 2.0 to the Quokka server.

    Generates QASM code to create quantum randomness.

    Uses Hadamard gates (h q[i]) to put qubits into superposition.

    If an API (my_quokka) is provided, it sends the QASM code for real execution.

    Otherwise, it returns the QASM code for manual execution.

    Finally, the qubits are measured, and the result is returned.

# Function to get quantum result


```python
# Function to get quantum result using QASM
def get_quokka_result(count, my_quokka=None):
    """Generates QASM code for quantum randomness and optionally sends it to an API."""
    qasm_code = f"""OPENQASM 2.0;
include "qelib1.inc";
qreg q[{count}];
creg c[{count}];
"""

    # Apply Hadamard gate to each qubit to generate superposition
    for i in range(count):
        qasm_code += f"h q[{i}];\n"

    # Measure each qubit into classical bits
    for i in range(count):
        qasm_code += f"measure q[{i}] -> c[{i}];\n"

    if my_quokka:
        try:
            response = requests.post(my_quokka, data={"qasm": qasm_code}, timeout=5, verify=False)
            if response.status_code == 200:
                return response.text.strip()[:count]  # Extract the first 'count' bits
        except requests.RequestException:
            pass  # If API fails, return QASM code for manual execution

    return qasm_code  # If no API is used, return the QASM code for external execution
```

## 6) Mapping Bitstring to Shot Directions

Converts 6-bit quantum output into goalkeeper's movement.

Ensures 1 or 2 unique moves, so the goalkeeper doesn’t move unrealistically.

Then we define function that takes the result(quantum) and maps it to corresponding directions for shooting

    map_to_symbols: This function takes the quantum result (a 6-bit string) and maps it to the corresponding shot directions using the symbol_map dictionary. It groups the bitstring into pairs of bits and converts them into the shot directions.

```python
# Convert quantum bits to goalkeeper dive positions
def map_to_symbols(bitstring):
    """Convert quantum bits into goalkeeper dive positions, ensuring valid movement."""
    if len(bitstring) < 6:
        bitstring = ''.join(random.choice("01") for _ in range(6))  # Generate fallback random bits

    keeper_moves = []
    for i in range(0, len(bitstring), 3):
        move = symbol_map.get(bitstring[i:i+3], random.choice(list(symbol_map.values())))  # Pick random valid move
        keeper_moves.append(move)

    # Return only 1 or 2 unique moves
    return list(set(keeper_moves[:2]))  # Ensures max 2 moves only


```

## 7) Evaluate the Shot Outcome

Now,for evaluating whether a chance took y a player is a goal or not,we define another function called evalute_spin which determines the outcome of a chance.They are instructed as follows:

✅ Same Shot & Dive Direction → Save

If the goalkeeper dives to the same location as the shot, the shot is saved instead of being a goal.

✅ If the shot is aimed at Center, it is saved only if the goalkeeper doesn’t dive

✅If the shot is aimed at Center, it is saved only if the goalkeeper didn’t dive to any other position.

```python
# Evaluate shot result
def evaluate_spin(keeper_symbols, player_shot):
    """Determines if the shot is saved or scores."""
    if player_shot in keeper_symbols:
        return "Saved!"
    elif player_shot == "Center" and all(pos not in keeper_symbols for pos in ["Top Left", "Bottom Left", "Top Right","Bottom Right"]):
        return "Saved!"
    return "Goal!"
```

## 8) Penalty function():

  We then define a function called take_penalty which handles the logic of the penalty taken by the end users

  

```python
# Take penalty shot
def take_penalty(choice):
    global current_player, player_shots

    # Stop if both players have finished
    if player_shots[0] >= max_attempts and player_shots[1] >= max_attempts:
        end_game()
        return

    # Show preparing image first
    with output:
        clear_output(wait=True)
        display(IPyHTML(f'<img src="{before_shot}" width="500">'))
        display(IPyHTML(f"<h2 style='color:white;'>Player {current_player + 1} is preparing to shoot...</h2>"))

    time.sleep(2)  # Simulate a delay before the shot result

    # Get quantum goalkeeper decision using QASM
    keeper_result = get_quokka_result(6)  # Fetches 6 quantum bits locally or via API
    keeper_symbols = map_to_symbols(keeper_result)

    # Player's shot choice
    player_shot = shot_directions[choice]
    result = evaluate_spin(keeper_symbols, player_shot)

    # Update scores
    if result == "Goal!":
        player_scores[current_player] += 1

    # Update shot count
    player_shots[current_player] += 1

    # Switch player
    current_player = 1 - current_player

    # Show result image
    result_image = goal_image if result == "Goal!" else miss_image

    with output:
        clear_output(wait=True)
        display(IPyHTML(f'<img src="{result_image}" width="500">'))
        display(IPyHTML(f"""
        <div style="text-align:center;">
            <h2 style='color:#FFFFFF;'>Player {current_player + 1}'s Turn ⚽</h2>
            <h3 style='color:#FFD700;'>Your Shot: {player_shot}</h3>
            <h3 style='color:#00FF00;'>Goalkeeper's Decision: {keeper_symbols}</h3>
            <h2 style='color:#FF4500;'>{result}</h2>
            <h2 style='color:#00FFFF;'>Score: Player 1 - {player_scores[0]} | Player 2 - {player_scores[1]}</h2>
            <h3 style='color:white;'>Attempts: Player 1 - {player_shots[0]} | Player 2 - {player_shots[1]}</h3>
        </div>"""))
```

The player selects a shot direction.

The goalkeeper’s decision is generated using quantum randomness.

The shot result (Goal! / Saved!) is displayed.

The player's shots are incremented for the current player and The shot direction is mapped using the player's choice (e.g., "Top Left", "Bottom Left", etc.).

It fetches the quantum result for the goalkeeper’s dive and maps it to the shot directions using the map_to_symbols() function.

The outcome of the shot is evaluated using the evaluate_spin() function

If the result is a goal, the current player's score is incremented.

The total attempts counter is incremented, and then it switches the turn to the other player and then the result image is chosen based on whether it was a goal or a miss.
The result image (either goal or miss) is displayed.

After that the scoreboard is updated with both players' scores and the number of attempts they've made.



If both players have finished their shots, it calls the end_game() function to end the game.

## 9) End Game Function

This function is created so s to declare the final result forthe match between player 1 and player 2
end_game(): This function evaluates the final result of the game:

It checks if Player 1’s score is higher than Player 2’s, and declares Player 1 the winner, or vice versa.

If both players have the same score, it declares the game a draw.

```python
# End game function
def end_game():
    winner = "Player 1" if player_scores[0] > player_scores[1] else "Player 2"
    if player_scores[0] == player_scores[1]:
        winner = "It's a draw!"

    with output:
        clear_output(wait=True)
        display(IPyHTML(f"<h1 style='color:#FF1493;'>Game Over! 🏆 {winner} Wins!</h1>"))
```
