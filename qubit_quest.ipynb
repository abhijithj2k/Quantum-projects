Your goal is to get to the **Quantum Gem**.

Move around using the right buttons and inspect adjacent walls with the left.

### Game Mechanics 🕹️

The game is your typical dungeon escape game where the players are to escape the dungeon while solving a series of quantum puzzles to escape. The game mecanics are outlined below:

1. **Movement**  
   - You are **Bunnyana Jones**!
   - Use the **keypad on the right of the screen** which is integrated with the UI to move cell by cell.  
    - Use the **keypad on the left of the screen** to inspect the wall according to the direction of the wall to you.
   - Classical walls ($\color{green}{\textsf{green}}$ viney wall) block physical movement; quantum walls (**hologram $\color{red}{\textsf{stop sign}}$**) may open or close based on your qubit measurements.
2. **Heads up!**
   - **Bombs** are dangerous! Be careful to not step on them!!
   - Some quantum walls are open to start with, but they can be a **trap**. With you be on the wall and some how the wall is turned on again (measured to `1`), then you will be crushed by the wall!

3. **Inspection & Editing**  
   - When adjacent to a quantum wall, switch to *editing mode* and drag gates from your toolbelt onto the cell’s gate slots.
   - Quantum gates on the wall with **$\color{red}{\textsf{red border}}$** are non-removable.
   - To interact with the quantum gates, simply **click on the gate**, then **click on where you want to put it**.
   - Each gate corresponds to an OpenQASM operation (e.g. `h`, `z`, `x`, `t`).

4. **Quantum Measurement**  
   Once you place gates, the game generates an OpenQASM script, sends it to the Quokka simulator, and updates each quantum wall’s state:

   ```
   OPENQASM 2.0;
   qreg q[N];
   creg c[N];

   <gate-sequence> q[i];

   measure q -> c;
   ```

   After this step, if the outcome of the classical bit matches a set of rules for the identity (i.e., `HZH=X`) the wall is removed.

  $$
|\text{initial state}\rangle → \fbox{cancelled out gates} → |\text{unchanged state}\rangle
$$

  •	Place/remove gates in editing mode by clicking a gate then clicking a slot.

## Quantum Computing Aspects ⚛️



### The Qubits 💾
Ah yes, the fundamental of Quantum Computing - **the qubit**. And of course, it will be the very first stepping stone for the quantum computing aspect of this game.
<br>

A single qubit's state can be represented as a **superposition** as follows:
$$|\psi\rangle = \alpha |0\rangle + \beta |1\rangle$$
<br>
- Where $|0\rangle$ and $|1\rangle$ are **orthogonal basis states**: $
|0\rangle = \begin{bmatrix} 1 \\ 0 \end{bmatrix},
|1\rangle = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$.

- $\alpha$ and $\beta$ are **complex numbers**, representing the **probability** of $|0\rangle$ and $|1\rangle$: $|\alpha|^2 + |\beta|^2 = 1$ (normalization).
- This makes our qubits to collapses to $|0\rangle$ with probability $|\alpha|^2$ or $|1\rangle$ with probability $|\beta|^2$.
<br><br>
This would make our game much more unpredictable and chaotic 😛😛.

### Hadamard Gate🚪
To control the qubit to exist in its superposition, the **Hadamard gate** is applied. The Hadamard gate is a fundamental **single-qubit** operation in quantum computing that creates superposition.
$$H = \frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix}$$

Applying the Hadamard gate to $|0\rangle$ or $|1\rangle$ would creates a perfect superposition:
$$H|0\rangle = \frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix} \begin{bmatrix}
1 \\
0
\end{bmatrix} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 \\
1
\end{bmatrix} =
\frac{|0\rangle + |1\rangle}{\sqrt{2}}$$
<br>
Here, $\alpha = \beta = \frac{1}{\sqrt{2}}$, and $\Big(\frac{1}{\sqrt{2}}\Big)^2 + \Big(\frac{1}{\sqrt{2}}\Big)^2 = 1$. So measuring would gives `0` or `1` with $50\%$ probability each.

$$H|1\rangle = \frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix} \begin{bmatrix}
0 \\
1
\end{bmatrix} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 \\
-1
\end{bmatrix} =
\frac{|0\rangle - |1\rangle}{\sqrt{2}}$$
<br>

Here, $\alpha = \beta = \frac{-1}{\sqrt{2}}$, and $\Big(\frac{-1}{\sqrt{2}}\Big)^2 + \Big(\frac{-1}{\sqrt{2}}\Big)^2 = 1$. So measuring would gives `0` or `1` with $50\%$ probability each. With each move of yours being measured, an $H$-gate can be unpredictable ...


### Pauli-X Gate🚪
When applied on to a qubit, the **Pauli-X gate** (or $X$-gate) flips that qubit's state (Yes! Just like a $NOT$ gate).


$$X = \begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}$$
If the qubit's state is $|0\rangle$:
$$X|0\rangle = \begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix} \begin{bmatrix}
1 \\
0
\end{bmatrix} = \begin{bmatrix}
0 \\
1
\end{bmatrix} = |1\rangle$$

If the qubit's state is $|1\rangle$:
$$X|1\rangle = \begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix} \begin{bmatrix}
0 \\
1
\end{bmatrix} = \begin{bmatrix}
1 \\
0
\end{bmatrix} = |0\rangle$$

The $X$-gate is integrated in virtually all quantum circuits and is used in various quantum algorithms. It would be a fundamental piece for you to open the gate!

### Pauli-Z Gate🚪

The Pauli-Z Gate (or $Z$-gate) will only flip the qubit's phase if it's state is $|1\rangle$. The qubit will remain unchanged if it is $|0\rangle$.

$$Z = \begin{bmatrix}
1 & 0 \\
0 & -1
\end{bmatrix}$$
If the qubit's state is $|0\rangle$:
$$Z|0\rangle = \begin{bmatrix}
1 & 0 \\
0 & -1
\end{bmatrix} \begin{bmatrix}
1 \\
0
\end{bmatrix} = \begin{bmatrix}
1 \\
0
\end{bmatrix} = |0\rangle$$

If the qubit's state is $|1\rangle$:
$$Z|1\rangle = \begin{bmatrix}
1 & 0 \\
0 & -1
\end{bmatrix} \begin{bmatrix}
0 \\
1
\end{bmatrix} = \begin{bmatrix}
0 \\
-1
\end{bmatrix} = - |1\rangle$$

Like the $X$-gate, the $Z$-gate would be a fundamental piece for you to solve the maze!

### Phase ($S$) Gate🚪

The Phase Gate (aka $S$-gate aka the $Z90$ gate aka the $\sqrt{Z}$ gate) represents a 90-degree rotation around the z-axis. This again only applies to the qubit if it's state is $|1\rangle$.

$$S = \begin{bmatrix}
1 & 0 \\
0 & i
\end{bmatrix}$$

$$S^2 = \begin{bmatrix}
1 & 0 \\
0 & i
\end{bmatrix} \times \begin{bmatrix}
1 & 0 \\
0 & i
\end{bmatrix} = \begin{bmatrix}
1 & 0 \\
0 & -1
\end{bmatrix}$$

If the qubit's state is $|0\rangle$:
$$S|0\rangle = \begin{bmatrix}
1 & 0 \\
0 & i
\end{bmatrix} \begin{bmatrix}
1 \\
0
\end{bmatrix} = \begin{bmatrix}
1 \\
0
\end{bmatrix} = |0\rangle$$

If the qubit's state is $|1\rangle$:
$$S|1\rangle = \begin{bmatrix}
1 & 0 \\
0 & i
\end{bmatrix} \begin{bmatrix}
0 \\
1
\end{bmatrix} = \begin{bmatrix}
0 \\
i
\end{bmatrix} $$

Solving with the $S$-gate may get a little tricky 😉.

````python
# @title The good ole send to quokka function
def send_to_quokka(program, count=1, my_quokka='quokka5'):  # my_quokka is a keyword argument with a default value
    # Step 1: Define the server address
    request_http = 'http://{}.quokkacomputing.com/qsim/qasm'.format(my_quokka)
    # Step 2: Prepare the data to send
    data = {
        'script': program,
        'count': count
    }

    # Step 3: Send the POST request
    result = requests.post(request_http, json=data, verify=False)
    # Step 4: Convert the response to a Python dictionary
    json_obj = False
    try: # if 502 or some other thing that happens that doesnt return json
      json_obj = json.loads(result.content)
    except Exception:
      print(result)

    try:
      # Return the result
      return json_obj['result']['c']
    except Exception:
      # gen a syntax error in the qasm so print out err msg

      if json_obj:
        print(json_obj)
      else:
        print("got bad data from quokka, could be offline")
````

````python
sprites = {
    "player":Image.from_file("/content/bunnyana_jones.png"),
    "goal":Image.from_file("/content/gem.png"),
    "quantumWallOn":Image.from_file("/content/quantumWallOn.png"),
    "quantumWallOff":Image.from_file("/content/quantumWallOff.png"),
    "horizWall":Image.from_file("/content/horizWall.png"),
    "vertWall":Image.from_file("/content/vertWall.png"),
    "dpad":Image.from_file("/content/dpad.png"),
    "close":Image.from_file("/content/close.png"),
    "bomb":Image.from_file("/content/bomb.png"),
}

class Button:
  def __init__(self, label, x, y, width, height, disabled=False):
    self.label = label
    self.x = x
    self.y = y
    self.w = width
    self.h = height

    self.pressed = False
    self.hovered = True
    self.prevDown = False

    self.disabled = disabled

    self.selected = False
    self.superSelected = False

  # Draws the button on the canvas, with special styling for exit, selected, or special buttons
  def draw(self):
    if self.label == "EXIT":
      canvas.draw_image(sprites["close"], self.x, self.y)
      return

    if self.selected or self.superSelected or (len(self.label) > 0 and self.label[0] == "_"):
      off = self.w*0.2
      drawRect(self.x-off/2,self.y-off/2,self.w+off,self.h+off, "red" if (len(self.label) > 0 and self.label[0] == "_") else "purple" if self.superSelected else "blue")

    drawRect(self.x,self.y,self.w,self.h, "red" if self.disabled else "#00ab83")
    canvas.text_align="center"
    drawText(self.label.strip("_"), self.x+self.w*0.6,self.y+self.h*0.4, col="white")
    canvas.text_align="start"

  # Updates the button's hover and pressed states based on mouse interaction
  def update(self):
    global mouse

    if (len(self.label) > 0 and self.label[0] == "_"):
      return
    self.hovered = AABBCollision(self.x,self.y,self.w,self.h,*mouse['pos'],0,0)
    self.pressed = mouse['pressed'] and self.hovered


@dataclass
class Cell:
  row: int
  col: int
  gates: list = field(default_factory=lambda: [None, None, None, None])
  measuredState: bool = False

@dataclass
class Level:
  playerPos: tuple
  goalPos: tuple
  toolbelt: list

  # eg: { (1,1): Cell }
  quantumWalls: dict # dict of cells

  # we can define walls as a pair of pairs:
  # eg: {((1,1), (1,2)): true} this represents a wall in the 1 row that is between col 1 and col 2
  walls: dict

  # eg: [(1,1), (2,2)]
  bombs: list

  rows: int = 5
  cols: int = 6

class Game:
  def __init__(self):
    self.running = True
    self.levels = []
    self.startingStateOfLevels = []
    self.currLevel = 0
    self.chosenGate = None

    self.measurementCache = []
    self.ordered_keys = []

    self.errorMsg = ""

    self.mazeW = canvas.width*0.75

    self.lossMsg = ""

    # move: standard movement
    # editing: pickup or place gates
    self.playerState = "move"
    self.prevState = None
    self.currInspectedCell = None

    self.skipUpdate = True

    gamepadCenter = (canvas.width*0.75, canvas.height*0.87)
    gamepadWH = [64,64]
    self.movementBtns = [
        Button("up", gamepadCenter[0]-gamepadWH[0]/2, gamepadCenter[1]-gamepadWH[1]*(3/2), *gamepadWH),
        Button("left", gamepadCenter[0]-gamepadWH[0]*(3/2), gamepadCenter[1]-gamepadWH[1]/2, *gamepadWH),
        Button("down", gamepadCenter[0]-gamepadWH[0]/2, gamepadCenter[1]+gamepadWH[1]*(1/2), *gamepadWH),
        Button("right", gamepadCenter[0]+gamepadWH[0]*(1/2), gamepadCenter[1]-gamepadWH[1]/2, *gamepadWH),
    ]

    gamepadCenter = (canvas.width*0.25, canvas.height*0.87)
    gamepadWH = [64,64]
    self.inspectButtons = [
        Button("up", gamepadCenter[0]-gamepadWH[0]/2, gamepadCenter[1]-gamepadWH[1]*(3/2), *gamepadWH, True),
        Button("left", gamepadCenter[0]-gamepadWH[0]*(3/2), gamepadCenter[1]-gamepadWH[1]/2, *gamepadWH, True),
        Button("down", gamepadCenter[0]-gamepadWH[0]/2, gamepadCenter[1]+gamepadWH[1]*(1/2), *gamepadWH, True),
        Button("right", gamepadCenter[0]+gamepadWH[0]*(1/2), gamepadCenter[1]-gamepadWH[1]/2, *gamepadWH, True),
    ]

    self.currLevelStruct = None

    self.lastSelectedCell = None
    self.lastSelectedCellType = None

    self.exitButton = Button("EXIT", 120, 48, 64,64)

  # Resets the current level to its initial state
  def resetCurrLevelStruct(self):
    self.currLevelStruct = copy.deepcopy(self.levels[self.currLevel])

  # Returns the current level structure
  def getCurrLevel(self):
    return self.currLevelStruct

  # Calculates and returns the maze dimensions based on the current level's rows and columns
  def getMazeSize(self):
    r =self.getCurrLevel().rows
    c =self.getCurrLevel().cols
    return (64*c, 96*r)

  # Generates a quantum circuit, sends it to Quokka for measurement, and caches results
  def measureQuantumWalls(self):
    level = self.getCurrLevel()
    quantumWalls = level.quantumWalls

    size = len(quantumWalls)
    qasm = f"OPENQASM 2.0;qreg q[{size}];creg c[{size}];"
    sdef = "gate s q {U(0,0,pi/2) q;}"
    qasm += sdef

    self.ordered_keys = list(quantumWalls.keys())
    for ind in range(len(self.ordered_keys)):
      wall = quantumWalls[self.ordered_keys[ind]]
      for gate in wall.gates:
        if gate:
          gateStr = gate.lower().strip("_")
          qasm += f"{gateStr} q[{ind}];"
    qasm += f"measure q -> c;"

    self.measurementCache = send_to_quokka(qasm, count=20) # 20 seems to work fine

    self.updateQuantumWalls()

  # Draws the maze background and border
  def drawMazeScreen(self, on = True):
    mWH = self.getMazeSize()

    outline = 32
    drawRect((canvas.width-mWH[0])/2-outline, (canvas.height*0.05)-outline,mWH[0] + outline*2, mWH[1]+outline*3,"gray")
    drawRect((canvas.width-mWH[0])/2, (canvas.height*0.05),*mWH,"black")
    drawRect((canvas.width)/2+2*mWH[0]/3-32*3, (canvas.height*0.05)+mWH[1]+outline*1.5-32,32,32,"green" if on else "red")
    drawText("bingus inc. ©",(canvas.width)/2-2*mWH[0]/3+32*3, (canvas.height*0.05)+mWH[1]+outline*1.5-32, 20, "#D3D3D3")

  # Draws the main game board with player, walls, bombs, and buttons
  def drawMainBoard(self):
    lvl = self.getCurrLevel()

    for btn in self.movementBtns:
      btn.update()

    canvas.draw_image(sprites["dpad"], self.movementBtns[0].x - self.movementBtns[0].w, self.movementBtns[0].y)

    for i in range(len(self.inspectButtons)):
      btn = self.inspectButtons[i]
      xOff = -1 if i == 1 else 1 if i == 3 else 0
      yOff = -1 if i == 0 else 1 if i == 2 else 0
      btn.disabled = not ((lvl.playerPos[0]+yOff, lvl.playerPos[1]+xOff) in lvl.quantumWalls)
      btn.update()
      btn.draw()

      if btn.pressed and not btn.disabled:
        self.playerState = "editing"
        self.currInspectedCell = lvl.quantumWalls[(lvl.playerPos[0]+yOff, lvl.playerPos[1]+xOff)]
        self.setToolbeltBtns()
        self.setCellBtns()

    classicalWallThickness = 16
    mWH = self.getMazeSize()

    self.drawMazeScreen()


    r = lvl.rows
    c = lvl.cols
    for row in range(r):
      for col in range(c):
        cellType = "empty"
        rect = [(canvas.width-mWH[0])/2 + col*(mWH[0]/c), (canvas.height*0.05) + row*(mWH[1]/r), mWH[0]/c, mWH[1]/r]

        if lvl.goalPos == (row, col):
          cellType = "goal"
          canvas.draw_image(sprites[cellType], rect[0],rect[1])
        if (row, col) in lvl.quantumWalls:
          cellType = "quantumWallOn" if lvl.quantumWalls[(row, col)].measuredState else "quantumWallOff"
          canvas.draw_image(sprites[cellType], rect[0],rect[1])
        if lvl.playerPos == (row, col):
          cellType = "player"
          canvas.draw_image(sprites[cellType], rect[0],rect[1])

        if lvl.bombs and (row, col) in lvl.bombs:
          cellType = "bomb"
          canvas.draw_image(sprites[cellType], rect[0],rect[1])

        # Draw walls
        if ((row, col), (row, col+1)) in lvl.walls or ((row, col+1), (row, col)) in lvl.walls: # vert
          canvas.draw_image(sprites["vertWall"], (canvas.width-mWH[0])/2 + (col+1)*(mWH[0]/c)-classicalWallThickness/2, (canvas.height*0.05) + row*(mWH[1]/r))
        if ((row, col), (row+1, col)) in lvl.walls or ((row+1, col), (row, col)) in lvl.walls: # horiz
          canvas.draw_image(sprites["horizWall"], (canvas.width-mWH[0])/2 + (col)*(mWH[0]/c)-classicalWallThickness/2, (canvas.height*0.05) + (row+1)*(mWH[1]/r)-classicalWallThickness/2)

  # Creates buttons for the toolbelt based on the current level's toolbelt items
  def setToolbeltBtns(self):
    gates = self.getCurrLevel().toolbelt
    w = 64
    h = 64
    size = 30
    center = canvas.width/2 - (len(gates)*(w*5/4))/2
    self.toolbeltBtns = []
    for i in range(len(gates)):
      x = center + i*(w*5/4)
      y = h*2
      self.toolbeltBtns.append(Button(gates[i], x, y, w, h, gates[i]==""))

  # Creates buttons for the gates in the currently inspected cell
  def setCellBtns(self):
    gates = self.currInspectedCell.gates
    w = 64
    h = 64
    size = 30
    center = canvas.width/2 - (len(gates)*(w*8/7))/2
    self.inspectedCellBtns = []
    for i in range(len(gates)):
      x = center + i*(w*8/7)
      y = h*6
      self.inspectedCellBtns.append(Button(gates[i], x, y, w, h, gates[i]==""))

  # Draws the editing interface with toolbelt and cell gate buttons
  def drawEditingBoard(self):
    w = 64
    h = 64
    size = 30
    extendedToolbeltRect = [0, h*1.75, canvas.width, h*1.5]

    self.drawMazeScreen()

    if self.lastSelectedCellType == "cell" and self.lastSelectedCell.label != "":
      drawRect((canvas.width-self.getMazeSize()[0])/2, h*1.75, self.getMazeSize()[0], h*1.5, "blue")

    drawText("TOOLBELT", canvas.width/2-size*2, h*2 - size*2, size, "white")
    for btn in self.toolbeltBtns:
      btn.draw()

    drawText("CELL", canvas.width/2-size, h*6-size*2, size, "white")
    for btn in self.inspectedCellBtns:
      btn.draw()

    self.exitButton.update()
    self.exitButton.draw()
    if self.exitButton.pressed:
      self.playerState = "move"
      self.measureQuantumWalls()

  # Draws the current game board based on the player's state
  def drawCurrBoard(self):
    match self.playerState:
      case "move":
        self.drawMainBoard()
      case "editing":
        self.drawEditingBoard()
      case _:
        print("Fatal Error: Invalid Player State", "Player state: ", self.playerState)

  # Checks if the player's movement is blocked by walls or boundaries
  def checkWallCollisions(self, updatedPlayerPos):
    playerPos = self.getCurrLevel().playerPos
    numOfRows = self.getCurrLevel().rows
    numOfCols = self.getCurrLevel().cols
    quantumWalls = self.getCurrLevel().quantumWalls
    walls = self.getCurrLevel().walls
    outerBounds = (updatedPlayerPos[0] < 0 or updatedPlayerPos[0] >= numOfRows or updatedPlayerPos[1] < 0 or updatedPlayerPos[1] >= numOfCols)
    wallCollision = ((updatedPlayerPos[0], updatedPlayerPos[1]), playerPos) in walls or (playerPos, (updatedPlayerPos[0], updatedPlayerPos[1])) in walls # check reverse order too
    quantumWallColls = quantumWalls.get(tuple(updatedPlayerPos), Cell(0,0,[])).measuredState
    return outerBounds or wallCollision or quantumWallColls

  # Determines which movement button or key is pressed
  def getMovementBtnsPressed(self):
    global lastKey
    btnsPressed = [i.pressed for i in self.movementBtns]
    movementMap = ["w","a","s","d"] # up left down right
    return movementMap[btnsPressed.index(True)] if True in btnsPressed else (lastKey if lastKey in movementMap else False)

  # Handles player movement based on input and collision checks
  def doPlayerMovement(self):
    key = self.getMovementBtnsPressed() #w/a/s/d or gamepad btn pressed
    if not key:
      return

    lvl = self.getCurrLevel()
    playerPos = lvl.playerPos
    updatedPlayerPos = [*playerPos]

    match key:
      case "w":
        updatedPlayerPos[0] -= 1
      case "a":
        updatedPlayerPos[1] -= 1
      case "s":
        updatedPlayerPos[0] += 1
      case "d":
        updatedPlayerPos[1] += 1

    if not self.checkWallCollisions(updatedPlayerPos):
      if (lvl.playerPos != (updatedPlayerPos[0], updatedPlayerPos[1])):
        self.updateQuantumWalls()
      lvl.playerPos = (updatedPlayerPos[0], updatedPlayerPos[1])

    if lvl.playerPos == lvl.goalPos:
      self.playerState = "win"

  # Updates the state of quantum walls based on cached measurement results
  def updateQuantumWalls(self):
    # if we send_to_quokka every moevement input it makes the game feel slow and unresponsive so to solve this:
    # when sending to quokka ask for n measurements and then save that to our cache and then just refresh cache whenever low
    # also make sure to refresh cache when updating the gates
    if self.measurementCache is None or len(self.measurementCache) < 2:
      self.measureQuantumWalls()

    for qubitNum in range(len(self.ordered_keys)):
      curr_key = self.ordered_keys[qubitNum]
      self.getCurrLevel().quantumWalls[curr_key].measuredState = int(self.measurementCache[-1][qubitNum]) == 1

    self.measurementCache.pop()

  # Updates the main game board, handling movement and loss conditions
  def updateMainBoard(self):
    self.doPlayerMovement()
    playerInWall = self.getCurrLevel().playerPos in self.getCurrLevel().quantumWalls and self.getCurrLevel().quantumWalls[self.getCurrLevel().playerPos].measuredState
    playerOnBomb = [self.getCurrLevel().playerPos == bombPos for bombPos in self.getCurrLevel().bombs]
    playerOnBomb = any(playerOnBomb)

    if playerInWall or playerOnBomb:
      self.playerState = "lose"
      if playerInWall:
        self.lossMsg = "You were crushed by a Quantum Wall!"
      else:
        self.lossMsg = "You stepped on a bomb!"

  # Resets the selected state of all toolbelt and cell buttons
  def setAllBtnsSelected(self, selected):
    for btn in self.inspectedCellBtns:
      btn.selected = selected
    for btn in self.toolbeltBtns:
      btn.selected = selected

  # Handles gate swapping and placement in the editing interface
  def updateEditingBoard(self):
    checkForSelectionCancel = mouse['down']
    prevSelection = self.lastSelectedCell

    for btn in self.toolbeltBtns:
      btn.update()
      if btn.pressed:
        for b in self.inspectedCellBtns:
          b.selected = True
        if self.lastSelectedCellType == "cell":
          self.currInspectedCell.gates[self.inspectedCellBtns.index(self.lastSelectedCell)] = btn.label
          if self.lastSelectedCell.label == "":
            del self.getCurrLevel().toolbelt[self.toolbeltBtns.index(btn)]
          else:
            self.getCurrLevel().toolbelt[self.toolbeltBtns.index(btn)] = self.lastSelectedCell.label

          self.setCellBtns()
          self.setToolbeltBtns()
          self.lastSelectedCellType = None
          self.setAllBtnsSelected(False)
          self.lastSelectedCell = None
          return

        btn.superSelected = True
        self.lastSelectedCell = btn
        self.lastSelectedCellType = "toolbelt"
    for btn in self.inspectedCellBtns:
      btn.update()
      if btn.pressed:
        self.setAllBtnsSelected(True)
        if self.lastSelectedCellType != None:
          if self.lastSelectedCellType == "toolbelt":
            if btn.label == "":
              del self.getCurrLevel().toolbelt[self.toolbeltBtns.index(self.lastSelectedCell)]
              self.currInspectedCell.gates[self.inspectedCellBtns.index(btn)] = self.lastSelectedCell.label
            else:
              self.getCurrLevel().toolbelt[self.toolbeltBtns.index(self.lastSelectedCell)] = btn.label
              self.currInspectedCell.gates[self.inspectedCellBtns.index(btn)] = self.lastSelectedCell.label
          else:
            self.currInspectedCell.gates[self.inspectedCellBtns.index(self.lastSelectedCell)] = btn.label
            self.currInspectedCell.gates[self.inspectedCellBtns.index(btn)] = self.lastSelectedCell.label

          self.setCellBtns()
          self.setToolbeltBtns()
          self.lastSelectedCellType = None
          self.lastSelectedCell = None
          return

        btn.superSelected = True
        self.lastSelectedCell = btn
        self.lastSelectedCellType = "cell"

    if checkForSelectionCancel and prevSelection == self.lastSelectedCell:
      # first check if we are placing in new
      # if we are adding a new one into the toolbelt
      h = 64
      extendedToolbeltRect = [0, h*1.75, canvas.width, h*1.5]
      if self.lastSelectedCellType == "cell" and self.lastSelectedCell.label != "" and AABBCollision(*mouse['pos'],0,0, *extendedToolbeltRect):
        self.getCurrLevel().toolbelt.append(self.currInspectedCell.gates[self.inspectedCellBtns.index(self.lastSelectedCell)])
        self.currInspectedCell.gates[self.inspectedCellBtns.index(self.lastSelectedCell)] = ""
        self.setCellBtns()
        self.setToolbeltBtns()

      for btn in self.toolbeltBtns:
        btn.superSelected = False
      for btn in self.inspectedCellBtns:
        btn.superSelected = False

      self.lastSelectedCell = None
      self.lastSelectedCellType = None
      self.setAllBtnsSelected(False)

  # Draws the appropriate game screen based on the game state (win, lose, or normal)
  def draw(self):
    if self.playerState == "DONE":
      self.drawMazeScreen(True)
      mWH = self.getMazeSize()
      drawRect((canvas.width-mWH[0])/2, (canvas.height*0.05),*mWH,"green")
      canvas.text_align="center"
      drawText("Well Done! You've completed the Game", canvas.width/2, canvas.height*0.3, 20, "white")
      drawText("Thanks for playing :)", canvas.width/2, canvas.height*0.32, 20, "white")
      canvas.text_align="start"
      return

    if self.playerState == "lose":
      self.drawMazeScreen(True)
      mWH = self.getMazeSize()
      drawRect((canvas.width-mWH[0])/2, (canvas.height*0.05),*mWH,"red")
      canvas.text_align="center"
      drawText(self.lossMsg, canvas.width/2, canvas.height*0.3, 20, "white")
      canvas.text_align="start"

      w = 256
      h = 64
      btn = Button("TRY AGAIN", (canvas.width-w)/2, canvas.height*0.35, w, h)
      btn.draw()
      btn.update()
      if btn.pressed:
        self.playerState = "move"
        for btn in self.inspectButtons:
          btn.pressed = False
        self.skipUpdate = True
        self.resetCurrLevelStruct()
        self.measureQuantumWalls()
      return
    if(self.playerState == "win"):
      self.drawMazeScreen(True)
      mWH = self.getMazeSize()
      drawRect((canvas.width-mWH[0])/2, (canvas.height*0.05),*mWH,"green")
      canvas.text_align="center"
      drawText("Well Done!", canvas.width/2, canvas.height*0.3, 20, "white")
      canvas.text_align="start"
      w = 256
      h = 64
      btn = Button("NEXT LEVEL", (canvas.width-w)/2, canvas.height*0.35, w, h)
      btn.draw()
      btn.update()
      if btn.pressed:
        self.playerState = "move"
        self.currLevel+=1
        if self.currLevel >= len(self.levels):
          self.playerState = "DONE"
          return
        for btn in self.inspectButtons:
          btn.pressed = False
        self.skipUpdate = True
        self.resetCurrLevelStruct()
        self.measureQuantumWalls()
      return
      return

    self.drawCurrBoard()

  # Updates the game state based on the current player state
  def update(self):
    if self.skipUpdate:
      self.skipUpdate = False
      return
    match self.playerState:
      case "move":
        self.updateMainBoard()
      case "editing":
        self.updateEditingBoard()
````

$$H^2 =  \frac{1}{\sqrt{2}} \begin{bmatrix}

\end{bmatrix} \frac{1}{\sqrt{2}} \times \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix} $$

$$ = \frac{1}{\sqrt{2}} \begin{bmatrix}
2 & 0 \\
0 & 2
\end{bmatrix} $$

$$= \begin{bmatrix}

0 & 1
